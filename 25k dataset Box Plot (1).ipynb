```python
import pandas as pd
import matplotlib.pyplot as plt

maindata = pd.read_csv("SOCR-HeightWeight.csv")

dataH = maindata["Height(Inches)"]
figH = plt.figure(figsize =(21, 9))
plt.boxplot(dataH, orientation='horizontal')
plt.show()

dataW = maindata["Weight(Pounds)"]
figW = plt.figure(figsize =(21, 9))
plt.boxplot(dataW, orientation='horizontal')
plt.show()

minH = dataH.min()
q1H  = dataH.quantile(0.25)
q2H  = dataH.median()        
q3H  = dataH.quantile(0.75)
maxH = dataH.max()
iqrH = q3H - q1H
print(f"Height - Min: {minH}, Max: {maxH} Q1: {q1H}, Q2: {q2H}, Q3: {q3H}, IQR: {iqrH}")

minW = dataW.min()
q1W  = dataW.quantile(0.25)
q2W  = dataW.median()
q3W  = dataW.quantile(0.75)
maxW = dataW.max()
iqrW = q3W - q1W
print(f"Height - Min: {minW}, Max: {maxW} Q1: {q1W}, Q2: {q2W}, Q3: {q3W}, IQR: {iqrW}")
```
